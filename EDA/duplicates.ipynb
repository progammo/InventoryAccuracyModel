```python
duplicates_df = df[df.duplicated(subset=['date', 'sku_id'], keep=False)]

# Sort them so duplicate pairs/groups sit next to each other for easy viewing
duplicates_df = duplicates_df.sort_values(by=['sku_id', 'date'])

print(f"Total duplicate rows found: {len(duplicates_df)}")
print(duplicates_df.head(20))
```
